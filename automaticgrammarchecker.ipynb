```python

!pip install language-tool-python

import language_tool_python

# Create grammar checker
tool = language_tool_python.LanguageTool('en-US')

print("====================================")
print("       AUTOMATIC GRAMMAR CHECKER")
print("====================================")

# Get input from user
text = input("\nEnter a sentence: ")

# Check grammar mistakes
matches = tool.check(text)

# Correct the sentence
corrected_text = language_tool_python.utils.correct(text, matches)

print("\nOriginal Text:", text)

if matches:
    print("\nGrammar Mistakes Found:")
    for mistake in matches:
        print("-", mistake.message)

    print("\nCorrected Text:", corrected_text)

else:
    print("\nNo grammar mistakes found!")
    print("Your sentence is correct.")

tool.close()
```
